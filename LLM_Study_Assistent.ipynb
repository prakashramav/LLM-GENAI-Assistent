#### Import Module from Gemini **Package**

```python
from google import genai
from google.colab import userdata
```

Use Client Class to Create an instance (for interacting with Gemini API)

```python
client = genai.Client(api_key = userdata.get("GEMINI_API_KEY"))
print(client)
```

Use Client object for communicating to Gemini LLM

```python
def study_assistent(user_prompt):
  response = client.models.generate_content(
      model = "gemini-2.5-flash",
      contents=user_prompt
  )
  return response
```

```python
output = study_assistent("Explain LLM Architecture")
print(output.text)
```
